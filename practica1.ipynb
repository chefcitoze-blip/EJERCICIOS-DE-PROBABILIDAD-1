##Experimento determinista

un experimento determinista es aquel que produce siemple el mismo resultado cuando se repite bajo las mismas condiciones iniciales

Ejemplo del area de un circulo

Un experimento determinista es aquel que produce siempre el mismo resultado cuando se repite bajo las mismas condiciones iniciales. Es decir, si las variables que intervienen en el experimento se mantienen constantes, el resultado será siempre el mismo.

Por ejemplo, en matemáticas, si evaluamos la operación 2+3, el resultado siempre será 5. De manera similar, si calculamos el área de un círculo usando la fórmula

$A=π⋅r^2$
y el radio
$r$ permanece constante, el resultado del cálculo será siempre el mismo.

```python
import sympy

def area(radio):
    return sympy.pi * (radio ** 2)
```

Otro ejemplo, si ejecutamos un programa que ordena una lista de números de mayor a menor, el resultado será siempre la misma lista ordenada siempre que los datos de entrada sean los mismos.

En todos estos casos, las mismas condiciones iniciales conducen inevitablemente al mismo resultado.

#Experimento aleatorio

En contraste, un experimento aleatorio es aquel que, aún cuando se repita bajo las mismas condiciones, no produce siempre el mismo resultado y no es posible predecir con certeza cuál ocurrirá.

Por ejemplo, al lanzar una moneda al aire no podemos saber con seguridad si el resultado será cara o cruz antes de observarlo.

```python
import random

moneda_lados = ["aguila", "sol"]

def lanzar_moneda():
    return random.choice(moneda_lados)

lanzar_moneda()
```

```python
import random

dado_lados = ["1", "2","3","4","5","6"]

def lanzar_dado():
    return random.choice(dado_lados)

lanzar_dado()
```

De manera similar, en un sorteo de lotería no es posible predecir con certeza cuál será el número ganador.

Aunque no podemos determinar con exactitud el resultado de un experimento aleatorio individual, sí podemos describir matemáticamente su comportamiento mediante probabilidades.
