## Probabilidad condicional

### Ejercicio 1
> Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salio un número par.

Espacio muestral: $Ω = ${1, 2, 3, 4, 5, 6}

$𝘈$: obtener un 2 $𝘈 = 1 $

$𝘉$: Obtener un número par $𝘉= 3$

$P(𝘉) = \frac{3}{6}$

$P(𝘈∩𝘉) =\frac{1}{6}$

$P(𝘈|𝘉) = $$\frac{\frac{1}{6}}{\frac{3}{6}}$$ = $$\frac{1}{3}$$ $

```python
dado = [1, 2, 3, 4, 5, 6]
A = [x for x in dado if x == 2]
B = [x for x in dado if x % 2 == 0]
interseccion = [x for x in A if x in B]
probabilidad = len(interseccion) / len(B)
print(f"Probabilidad de obtener un 2 dado que salio un numero par: {probabilidad:.4f}")
print(f"Porcentaje:{probabilidad * 100:.2f}%")
```

### Ejercicio 2
> Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

$P(𝘈|𝘉) = $$\frac{18}{120}$$ = $$\frac{3}{20}$$ = 0.15$



```python
from itertools import permutations
dado = [1, 2, 3, 4, 5, 6]

S = list(permutations(dado, 3))

P = [x for x in S if sum(x) == 9]

probabilidad = len(P) / len(S)

print(f"Probabilidad de que la suma sea 9 dado que las tres caras son distintas: {probabilidad:.2f}")
print(f"Porcentaje:{probabilidad * 100:.2f}%")
```

## Teorema de probabilidad total

### Ejercicio 1
> El 70% de los clientes ven un anuncio y, de ellos, compra el 30%. El 30% no ve el anuncio y, de ellos, compra el 10%. Calcula la probabilidad total de que un cliente compre.

$P(A) = 0.7$, $P(C|A) = 0.3$, $P(B) = 0.3$ y $P(C|B) = 0.1$

$P(C) = 0.3(0.7) + 0.1(0.3) = 0.21 + 0.03 = 0.24 $

La probabilidad de que un cliente compre es del **24%**

```python
p_compra = (0.3 * 0.7 + 0.1 * 0.3)
print(f"Probabilidad total de que un cliente compre: {p_compra:.2f}")
print(f"Porcentaje:{p_compra * 100:.2f}%")
```

## Teorema de Bayes

### Ejercicio 1
> El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

$P(S) = 0.2$, $P(O|S) = 0.8%$, $P(D) = 0.8$ y $P(O|D) = 0.1$

$P(O) = (0.8)(0.2) + (0.1)(0.8) = 0.16 + 0.08 = 0.24$

$P(S|O) = $$\frac{(0.8)(0.2)}{0.24}$$ = $$\frac{0.16}{0.24}$$ ≈ 0.6667 $

La probabilidad de que sea spam es alrededor del **66.67%**

```python
p_spam = 0.2
p_oferta_spam = 0.8
p_deseado = 0.8
p_oferta_deseado = 0.1

p_oferta = (p_oferta_spam * p_spam) + (p_oferta_deseado * p_deseado)
p_spam_oferta = (p_oferta_spam * p_spam) / p_oferta

print(f"Probabilidad de que sea spam: {p_spam_oferta:.4f}")
print(f"Porcentaje:{p_spam_oferta:.2%}")
```

### Ejercicio 2
> En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

>Si se registró una venta, ¿qué navegador es más probable que haya usado el cliente?

$C$: Chrome, $S$: Safari, $N$: Otros y $B$: Compra

$P(C ∩ B) = 0.03 * 0.40 = 0.012$

$P(S ∩ B) = 0.04 * 0.35 = 0.014$

$P(N ∩ B) = 0.02 * 0.25 = 0.005$

$P(B) = 0.012 + 0.014 + 0.005 = 0.031$

$P(C|B) = $$\frac{0.012}{0.031}$$ ≈ 0.3871$

$P(S|B) = $$\frac{0.014}{0.031}$$ ≈ 0.4516$

$P(N|B) = $$\frac{0.005}{0.031}$$ ≈ 0.1613$

Es más probale que haya usado Safari con alrededor de **45.16%**

```python
navegadores = {
    "Chrome": {"porcentaje": 0.40, "compra": 0.03},
    "Safari": {"porcentaje": 0.35, "compra": 0.04},
    "Otros": {"porcentaje": 0.25, "compra": 0.02}
}

probabilidad_compra = {}

for navegador in navegadores:
    probabilidad_compra[navegador] = (
        navegadores[navegador]["porcentaje"] * navegadores[navegador]["compra"]
    )

p_compra = sum(probabilidad_compra.values())

probabilidad_navegador = {}

for navegador in navegadores:
    probabilidad_navegador[navegador] = (
        probabilidad_compra[navegador] / p_compra
    )
print("Probabilidad del navegador dado una venta:")
for navegador, probabilidad in probabilidad_navegador.items():
    print(f"{navegador}: {probabilidad:.2%}")
mas_probable = max(probabilidad_navegador, key=probabilidad_navegador.get)
print(f"El navegador más probable es: {mas_probable}")
```
