# Enfoque frecuentista de probabilidad

## Ejercicio 1
> Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.

Probabilidad clásica: $P(𝘈) = \frac{1}{9} ≈ 0.1111 $

```python
import random as rd
import matplotlib.pyplot as plt

def lanzar_dado():
  return rd.randint(1,6)

def lanzar_dos_dados():
  return lanzar_dado(), lanzar_dado()

n = 10000

contar_casos = 0

frecuencias = []
for i in range(1, n + 1):
  dado1, dado2 = lanzar_dos_dados()
  if dado1 + dado2 == 5:
    contar_casos += 1
  frecuencia = contar_casos / i
  frecuencias.append(frecuencia)

print("Número de lanzamientos:", n)
print("Probabilidad de que la suma de los dados sea exactamente 5:", frecuencia)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(frecuencias, label="Frecuencia relativa de la suma exacta de 5")
plt.axhline(y=0.1111, color='r', linestyle='--', label="Probabilidad teorica = 1/9")
plt.xlabel("Número de lanzamientos")
plt.ylabel("Frecuencia relativa")
plt.title("Convergencia de la probabilidad empírica al valor teórico")
plt.legend()
plt.show()
```

## Ejercicio 2
> Simula 10 000 PIN de 4 digitos, del 0 al 9, y estima la probabilidaad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

Probabilidad clásica: $P(𝘈) = $$\frac{18}{25} = 0.72$

```python
def generar_pin(digitos):
  return [rd.randint(0,9) for _ in range(digitos)]
n = 10000
conteo = 0
frecuenciasp = []

for i in range(1, n + 1):
  pin = generar_pin(4)
  if pin[0] != pin[1] != pin[2] != pin[3]:
    conteo += 1

  frecuenciasp.append(conteo/ i)

print("Número de PINs generados:", n)
print("Probabilidad de que ninguno se repita:", conteo/n )
```

```python
plt.figure(figsize=(10, 6))
plt.plot(frecuenciasp, label="Frecuencia relativa de que ningun número se repita")
plt.axhline(y=0.72, color='r', linestyle='--', label="Probabilidad teorica = 18/25")
plt.xlabel("Número de contraseñas")
plt.ylabel("Frecuencia relativa")
plt.title("Cuatro digitos distintos se acercan a 18/25")
plt.legend()
plt.show()
```

## Ejercicio 3
> Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

$𝘈 = 9 , 𝘉 = 10$

Probabilidad clásica de A: $P(𝘈) = $$\frac{25}{216}$$ ≈ 0.1157 $

Probabilidad clásica de B: $P(𝘉) = $$\frac{27}{216}$$ = 0.125 $

```python
def lanzar_tres_dados():
  return lanzar_dado(), lanzar_dado(), lanzar_dado()

n = 20000
conteo_9 = 0
frecuencia_9 = []
conteo_10 = 0
frecuencia_10 = []

for i in range(1, n + 1):
  dado1, dado2, dado3 = lanzar_tres_dados()
  suma = dado1 + dado2 + dado3
  if suma == 9:
    conteo_9 += 1
    frecuencia_9.append(conteo_9/i)
  elif suma == 10:
    conteo_10 += 1
    frecuencia_10.append(conteo_10/i)

print("Número de lanzamientos:", n)
print("Probabilidad de que la suma sea 9:", conteo_9/n)
print("Probabilidad de que la suma sea 10:", conteo_10/n)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(frecuencia_9, label="Frecuencia relativa de que la suma sea 9")
plt.plot(frecuencia_10, label="Frecuencia relativa de que la suma sea 10")
plt.axhline(y=0.1157, color='r', linestyle='--', label="Probabilidad teorica que sea 9 = 25/216")
plt.axhline(y=0.125, color='g', linestyle='--', label="Probabilidad teorica que sea 10 = 27/216")
plt.xlabel("Lanzamientos")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion de resultados")
plt.legend()
plt.show()
```

## Ejercicio 4
> Basado en el ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y  calcula la probabilidad de obtener una azul

```python
def sacar_bola():
  urna = ["rojo", "rojo", "rojo", "azul", "azul"]
  return rd.choice(urna)

n = 2000
conteo_azul = 0
frecuencia_azul = []

for i in range(1, n + 1):
  resultado = sacar_bola()
  if resultado == "azul":
    conteo_azul += 1
  frecuencia = conteo_azul/i
  frecuencia_azul.append(frecuencia)

print("Número de extracciones:", n)
print("Probabilidad de obtener una bola azul:", conteo_azul/n)
```

```python
plt.figure(figsize=(10, 6))
plt.plot(frecuencia_azul, label="Frecuencia relativa de que sea azul")
plt.axhline(y=0.4, color='r', linestyle='--', label="Probabilidad teorica = 2/5")
plt.xlabel("Extracciones")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion de resultados")
plt.legend()
plt.show()
```

##Ejercicio 5
> Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

```python
def lanzar_moneda():
  moneda = ["aguila","sol"]
  return rd.choice(moneda)

def serie_volados(volados = 20, aguilas_seguidas = 5):
  conteo_aguilas = 0
  for _ in range(volados):
    resultado = lanzar_moneda()
    if resultado == "aguila":
      conteo_aguilas += 1
      if conteo_aguilas == aguilas_seguidas:
        return True
    else:
      conteo_aguilas = 0
  return False

n = 10000
conteo_volados = 0
frecuencia_volados = []

for i in range(1, n + 1):
  if serie_volados():
    conteo_volados += 1
  frecuencia = conteo_volados/i
  frecuencia_volados.append(frecuencia)

print("Número de series de volados:", n)
print("Probabilidad de que aparezcan 5 águilas seguidas:", conteo_volados/n)
```

```python
plt.figure(figsize = (10,6))
plt.plot(frecuencia_volados, label = "Frecuencia relativa de que aparezcan 5 águilas seguidas")
plt.axhline(y=0.249, color = 'r', linestyle = '--', label = "Probabilidad teorica ≈ 0.249")
plt.xlabel("Series de volados")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion de resultados")
plt.legend()
plt.show()
```

## Ejercicio 6 (bonus)
> Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
n_clientes = 500
umbral_espera = 10

tiempo_reloj = 0
cajero_libre = 0

conteo_espera = 0
frecuencia_espera = []
tiempos_espera = []

for i in range(1, n_clientes + 1):

    tiempo_llegadas = rd.randint(1, 5)
    tiempo_reloj += tiempo_llegadas
    llegada_cliente = tiempo_reloj

    inicio_servicio = max(llegada_cliente, cajero_libre)

    espera = inicio_servicio - llegada_cliente
    tiempos_espera.append(espera)

    tiempo_servicio = rd.randint(2, 4)
    cajero_libre = inicio_servicio + tiempo_servicio

    if espera > umbral_espera:
        conteo_espera += 1

    frecuencia_espera.append(conteo_espera/ i)

print("Clientes totales:", n_clientes)
print("Clientes que esperaron más de 10 min:", conteo_espera)
print("Frecuencia de que la espera sea mayor a 10 min:", conteo_espera/n_clientes)
```

```python
plt.figure(figsize = (10,6))
plt.plot(frecuencia_espera, label = "Frecuencia relativa de que la espera sea mayor a 10 min")
plt.axhline(y=0.29, color = 'r', linestyle = '--', label = "Probabilidad teorica ≈ 0.294")
plt.xlabel("Clientes")
plt.ylabel("Frecuencia relativa")
plt.title("Comparacion de resultados")
plt.legend()
plt.show()
```
